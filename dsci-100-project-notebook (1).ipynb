```python
# Calculate the relative proportion of each category
home_plot = alt.Chart(tidy_march_madness).mark_bar().encode(
    x=alt.X("first_game_at_home:N").title("First Tournament Game at Home (Y/N)"),
    y=alt.Y("count():Q").title("Number of Teams")
).properties(
    title="First Tournament Game Location")

# Display the bar chart
home_plot
```

This bar chart compares the average total win percentage of teams that played their first tournament game at home (Y) and teams that did not (N). By comparing the two averages, we can see whether teams playing at home tend to have a higher overall win percentage. This helps us explore whether first game location could be useful for predicting a team's overall performance. 

```python
import altair as alt

# Calculate the average win percentage for each group
average_wins = tidy_march_madness.groupby("first_game_at_home").agg(
    mean_win_pct=("total_wins_pct", "mean")
).reset_index()

# Create a bar chart comparing average win percentages
average_plot = alt.Chart(average_wins).mark_bar().encode(
    x=alt.X("first_game_at_home:N").title("First Tournament Game at Home (Y/N)"),
    y=alt.Y("mean_win_pct:Q").title("Average Total Win Percentage")
).properties(
    title="Average Win Percentage by First Game Location")

# Display the bar chart
average_plot
```

This scatter plot compares the total win percentages of 50 randomly selected teams based on whether they played their first tournament game at home. Green points represent teams that played at home, while red points represent teams that did not. This graph helps us visually compare win percentages between the two groups and explore whether playing at home is associated with a higher overall win percentage. Random sampling also keeps the graph readable by reducing overcrowding.

```python
# Randomly select 50 teams
random_teams = march_madness.sample(n=50)

# Create scatter plot
team_plot = alt.Chart(random_teams).mark_point(
    filled=True
).encode(
    x=alt.X("school:N").axis(labelAngle=-90).title("Team Name"),
    y=alt.Y("total_wins_pct:Q").title("Total Win Percentage"),
    color=alt.Color("first_game_at_home:N").scale(
        domain=["Y", "N"],
        range=["green", "red"]
    ).title("First Game at Home")
).properties(
    title="Total Win Percentage of 50 Random Teams",
)

team_plot
```
